````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 3, 6, 8]
y2 = [5, 3, 6, 4, 7]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 4))

ax1.plot(x, y1, marker='o')
ax1.set_title("Line Plot")
ax1.set_xlabel("X Values")
ax1.set_ylabel("Y Values")

ax2.scatter(x, y2, color='green')
ax2.set_title("Scatter Plot")
ax2.set_xlabel("X Values")
ax2.set_ylabel("Y Values")

fig.suptitle("Working with Figure and Axes")
plt.tight_layout()
plt.show()
````

````python

import numpy as np
from bokeh.core.enums import Anchor, ImageOrigin
from bokeh.models import Select
from bokeh.plotting import figure, show
from bokeh.layouts import column
# Create a small 2 × 2 colored image
colors = [
    [0xff9a4b36, 0xffe5d2a5],
    [0xff72c0fd, 0xff2600a5]
]
img = np.array(colors, dtype=np.uint32)
# Create the plot
p = figure(
    title="Image Origin and Anchor",
    x_range=(-10, 10),
    y_range=(-10, 10),
    width=500,
    height=400
)
# Place the image at (0, 0)
image = p.image_rgba(
    image=[img],
    x=0, y=0,
    dw=8, dh=8
)
# Mark the reference point
p.scatter([0], [0], size=10, color="black")
# Dropdown to change image placement
anchor = Select(
    title="Anchor",
    options=list(Anchor),
    value=image.glyph.anchor
)
anchor.js_link("value", image.glyph, "anchor")

# Dropdown to change image orientation
origin = Select(
    title="Origin",
    options=list(ImageOrigin),
    value=image.glyph.origin
)
origin.js_link("value", image.glyph, "origin")

show(column(p, anchor, origin))
````

````python
from bokeh.plotting import figure, show

p = figure(
    title="Bokeh Scatter Plot: Different Markers",
    x_axis_label="X Values",
    y_axis_label="Y Values",
    width=600,
    height=400
)

p.scatter([1, 2, 3], [3, 5, 4],
          marker="circle", size=18,
          color="blue", legend_label="Circle")

p.scatter([4, 5, 6], [4, 3, 5],
          marker="triangle", size=18,
          color="red", legend_label="Triangle")

p.scatter([7, 8, 9], [5, 4, 3],
          marker="square", size=18,
          color="green", legend_label="Square")

p.legend.location = "top_left"

show(p)
````

````python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

g = sns.displot(
    random.exponential(scale=2, size=1000),
    kind="kde",
    height=4,
    aspect=1.5
)

g.set_axis_labels("Waiting Time (Time Units)", "Probability Density")

g.fig.suptitle(
    "Exponential Distribution (Scale = 2)",
    fontsize=16,
    y=1.02
)

plt.show()
````

````python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(
    random.chisquare(df=2, size=1000),
    kind="kde",
    height=4,
    aspect=1.5
)

plt.xlabel("Chi-Square Value")
plt.ylabel("Probability Density")
plt.title("Chi-Square Distribution (df = 2)")

plt.show()
````

````python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

for df in [1, 2, 5, 10]:
    sns.kdeplot(
        random.chisquare(df=df, size=1000),
        label=f"df = {df}"
    )

plt.xlabel("Chi-Square Value")
plt.ylabel("Probability Density")
plt.title("Effect of Degrees of Freedom on Chi-Square Distribution")
plt.legend()

plt.show()
````

````python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

g = sns.displot(
    random.pareto(a=2, size=1000),
    kind="kde",
    height=4,
    aspect=1.5
)

g.set_axis_labels(
    "Pareto-Distributed Quantity",
    "Probability Density"
)

g.fig.suptitle(
    "Pareto Distribution (a = 2)",
    fontsize=16,
    y=1.02
)

plt.show()
````

````python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

x = random.zipf(a=2, size=1000)
g = sns.displot(
    x[x < 10],
    height=4,
    aspect=1.5
)

g.set_axis_labels(
    "Rank / Value",
    "Frequency"
)
g.fig.suptitle(
    "Zipf Distribution (a = 2)",
    fontsize=16,
    y=1.02
)
plt.show()
````

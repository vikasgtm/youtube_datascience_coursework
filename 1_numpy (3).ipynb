We can use op_dtypes argument and pass it the expected datatype to change the datatype of elements while iterating like int to string.

NumPy does not change the data type of the element in-place (where the element is in array)
and in order to enable it in nditer() we pass flags.

```python
arr = np.array([1, 2, 3])

for x in np.nditer(arr, flags=['buffered'], op_dtypes=['S']):
  print(x)
```

```python
# Create a filter array that will return only values higher than 42:

arr = np.array([41, 42, 43, 44])

# Create an empty list
filter_arr = []

# go through each element in arr
for element in arr:
  # if the element is higher than 42, set the value to True, otherwise False:
  if element > 42:
    filter_arr.append(True)
  else:
    filter_arr.append(False)

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

```python
# Create a filter array that will return only even elements from the original array:

arr = np.array([1, 2, 3, 4, 5, 6, 7])

# Create an empty list
filter_arr = []

# go through each element in arr
for element in arr:
  # if the element is completely divisble by 2, set the value to True, otherwise False
  if element % 2 == 0:
    filter_arr.append(True)
  else:
    filter_arr.append(False)

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

```python

# create numpy 1d-array
arr = np.array([2, 0, 1, 5,
				4, 1, 9])

print("Given array:", arr)

# sort an array in
# ascending order

# np.argsort() return
# array of indices for
# sorted array
sorted_index_array = np.argsort(arr)

# sorted array
sorted_array = arr[sorted_index_array]

print("Sorted array:", sorted_array)

# we want 1 largest value
n = 1

# we are using negative
# indexing concept

# take n largest value
rslt = sorted_array[-n : ]

# show the output
print("{} largest value:".format(n),
	rslt[0])
```

```python

# we want 3 largest value
n = 3

# we are using negative
# indexing concept

# find n largest value
rslt = sorted_array[-n : ]

# show the output
print("{} largest value:".format(n),
	rslt)
```

```python
# create a 2D array
c = np.array([[12, 5, np.nan, 7],
                 [2, 61, 1, np.nan],
                 [np.nan, 1,
                  np.nan, 5]])

# remove nan values using numpy.isnan()
# and numpy.logical_not
d = c[~(np.isnan(c))]

# print the results
print("Original 2D array   ")
print(c)
print()

print("2D array converted to 1D after removing nan values   ")
print(d)
```
